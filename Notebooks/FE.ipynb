```python
# Beispiel: alle eindeutigen Clubs abrufen
clubs = joined["Club"].unique()

# Für jeden Club einen DataFrame erstellen
dfs = {club: joined[joined["Club"] == club] for club in clubs}

# Zugriff auf einen DataFrame für z.B. Club "Bayern"
america = dfs["Club América"]
manu = dfs["Manchester United"]
spurs = dfs["Tottenham Hotspur"]
ol = dfs["Olympique Lyon"]
bayern = dfs["Bayern München"]
napoli = dfs["Napoli"]
parisfc = dfs["Paris FC"]
chelsea = dfs["Chelsea"]
arsenal = dfs["Arsenal"]
cor = dfs["Corinthians"]
juve = dfs["Juventus"]
wolfsburg = dfs["Wolfsburg"]
```

```python
import pandas as pd

# Beispiel: Perzentile berechnen (z.B. 33% und 66%)
low_threshold = joined['Comment Like Count'].quantile(0.33)
high_threshold = joined['Comment Like Count'].quantile(0.66)

def classify_popularity(count):
    if count <= low_threshold:
        return 'Low'
    elif count <= high_threshold:
        return 'Average'
    else:
        return 'High'

# Liste aller DataFrames
dfs = [america, manu, spurs, chelsea, arsenal, bayern, wolfsburg, juve, napoli, cor, ol, parisfc]

# Sicherstellen, dass alle DataFrames Kopien sind
dfs = [df.copy() for df in dfs]

# Neue Spalte mit .loc und apply hinzufügen
for df in dfs:
    df.loc[:, 'Comment Popularity'] = df['Comment Like Count'].apply(classify_popularity)

```
