```python
import pandas as pd
df = pd.read_csv("../Datasets/unified.csv")
                 
```

```python


# Step 1: Select only two columns
df_two_cols = df[["Video Title", "Gender of Sport"]]

# Step 2: Drop duplicates only on Video Title
df_distinct = df_two_cols.drop_duplicates(subset=["Video Title"]).reset_index(drop=True)

# Step 3: Manual review loop
for i, row in df_distinct.iterrows():
    print(f"\nVideo Title: {row['Video Title']} | Current Gender: {row['Gender of Sport']}")
    choice = input("Do you want to toggle the gender? (y/n): ").strip().lower()
    if choice == "y":
        df_distinct.at[i, "Gender of Sport"] = "Female" if row["Gender of Sport"] == "Male" else "Male"

print("\nFinal reviewed DataFrame:")
display(df_distinct)
```
